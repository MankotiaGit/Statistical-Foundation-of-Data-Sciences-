```python
import pandas as pd
import numpy as np
from scipy.stats import norm
```

```python
url = "https://vincentarelbundock.github.io/Rdatasets/csv/AER/TeachingRatings.csv"

df = pd.read_csv(url)
```

(Q1)

```python
total = len(df)

count_greater_45 = len(df[df['eval'] > 4.5])

probability_q1 = count_greater_45 / total

print("\nQ1: Probability of evaluation score greater than 4.5")
print("Count:", count_greater_45)
print("Probability:", probability_q1)
print("Percentage:", probability_q1 * 100, "%")
print("Probability percentage:", probability_q1 * 100, "%")
```

(Q2)

```python
count_between = len(df[(df['eval'] > 3.5) & (df['eval'] < 4.2)])

probability_q2 = count_between / total

print("\nQ2: Probability of evaluation score greater than 3.5 and less than 4.2")
print("Count:", count_between)
print("Probability:", probability_q2)
print("Percentage:", probability_q2 * 100, "%")
```

(Q3)

```python
population_mean = 12
population_std = 5.5
sample_size = 36
sample_mean = 10.7


standard_error = population_std / np.sqrt(sample_size)


z_score = (sample_mean - population_mean) / standard_error


p_value = 2 * norm.cdf(-abs(z_score))

print("\nQ3: Two-Tailed Z-Test")

print("Null Hypothesis (H0): Mean = 12")
print("Alternative Hypothesis (H1): Mean != 12")

print("Standard Error:", standard_error)
print("Z-score:", z_score)
print("P-value:", p_value)


alpha = 0.05

if p_value < alpha:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: The mean is significantly different from 12.")
else:
    print("Decision: Fail to Reject the Null Hypothesis")
    print("Conclusion: There is no significant evidence that the mean is different from 12.")
```
