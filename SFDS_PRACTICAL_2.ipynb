Therefore, it is cross-sectional rather than time-series data.



**(Q2)**

```python
students = df["students"]
print("Mean =", students.mean())
print("Median =", students.median())
print("Minimum =", students.min())
print("Maximum =", students.max())
```

In this snippet I extracted the students column from the DataFrame and computed its mean median minimum and maximum values using built-in statistical functions before printing each result

**(Q4)**

```python
import matplotlib.pyplot as plt
beauty = df['beauty']
print(beauty.describe())
print("Skewness:", beauty.skew())
plt.figure(figsize=(8, 5))
plt.hist(beauty, bins=12, edgecolor='black', color='steelblue')
plt.title("Distribution of Beauty Ratings (TeachingRatings dataset)")
plt.xlabel("Beauty score (standardized)")
plt.ylabel("Number of instructors")
plt.axvline(beauty.mean(), color='red', linestyle='--', label=f"Mean = {beauty.mean():.2f}")
plt.legend()
plt.tight_layout()
plt.savefig("beauty_histogram.png", dpi=150)
plt.show()
```

The histogram of beauty scores appears to be roughly unimodal. The distribution is slightly right-skewed, as there is a longer tail towards higher beauty scores, with a few observations above 1.0 and up to around 2.0. There are also some lower scores extending to about-1.5. Thus, the data is not perfectly symmetric and shows a few extreme values at both ends, particularly on the right side.

**(Q5)**

```python
df.groupby("gender")["beauty"].mean()
```

The average beauty score for female instructors is 0.1161. The average beauty score for male instructors is -0.0845. Since 0.1161 > -0.0845, female instructors have a slightly higher average beauty score than male instructors in this dataset.

```python
df.groupby("gender")["beauty"].std()
```

The standard deviations are: Female: 0.8178 Male: 0.7571. This indicates that beauty scores among female instructors are slightly more spread out (more variable) than those among male instructors.

**(Q6)**

```python
pd.crosstab(df["gender"], df["tenure"])
```

```python
pd.crosstab(
    df["gender"],
    df["tenure"],
    normalize="index"
) * 100
```

Male professors with tenure: 80.60% Female professors with tenure: 74.36% To determine whether tenure status differs by gender, a cross-tabulation of gender and tenure status was performed. The results show that 74.36% of female instructors and 80.60% of male instructors are tenured professors. Since the percentage of tenured male instructors is higher than that of female instructors, tenure status appears to differ somewhat by gender, with males having a higher tenure rate in this dataset. However, a statistical test such as the Chi-Square test would be required to determine whether this difference is statistically significant.
