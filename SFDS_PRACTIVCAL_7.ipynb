```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy import stats
```

```python
url = "https://vincentarelbundock.github.io/Rdatasets/csv/AER/TeachingRatings.csv"
df = pd.read_csv(url)
```

```python
df['female'] = (df['gender'] == 'female').astype(int)
```

```python
df.head()
```

(Q1) Regression with T-test

```python
X = sm.add_constant(df['female'])
y = df['eval']

model1 = sm.OLS(y, X).fit()
print(model1.summary())
```

(Q2) Regression with Anova

```python
df['age_group'] = pd.cut(df['age'], bins=3, labels=['Young', 'Middle', 'Old'])

model2 = smf.ols('beauty ~ age_group', data=df).fit()

anova_table = sm.stats.anova_lm(model2, typ=2)
print(anova_table)
```

(Q3) Correlation

```python
X = sm.add_constant(df['beauty'])
y = df['eval']

model3 = sm.OLS(y, X).fit()
print(model3.summary())
corr, p_value = stats.pearsonr(df['beauty'], df['eval'])
print(f"\nCorrelation coefficient: {corr:.4f}, p-value: {p_value:.6f}")
```
