PART A: Implement Metropolis Algorithm

```python

import random
import math

# Define the function
def energy(x):
    return x**2

# Initial value
x = 5
temperature = 1.0
iterations = 1000

# Metropolis Algorithm
for i in range(iterations):
    new_x = x + random.uniform(-1, 1)

    delta = energy(new_x) - energy(x)

    if delta < 0:
        x = new_x
    else:
        probability = math.exp(-delta / temperature)

        if random.random() < probability:
            x = new_x

print("Final value of x:", round(x, 4))
print("Minimum energy:", round(energy(x), 4))
```

The Metropolis Algorithm is a probabilistic algorithm used to find the minimum value of a function. It generates a new random value and accepts it if the energy decreases. If the energy increases, it may still accept the value based on a probability.
In this, I implemented the Metropolis Algorithm to find the minimum value of a function using random values and probability.

PART B: Deterministic Model


```python

# Initial population
population = 100

# Growth rate
growth_rate = 0.10

# Number of time steps
time_steps = 10

print("Year 0: Population =", population)

# Deterministic Model
for year in range(1, time_steps + 1):
    population = population + (growth_rate * population)

    print("Year", year, ": Population =", round(population, 2))
```

A Deterministic Model is a mathematical model in which the output is determined by fixed inputs and rules. In this program, population increases by 10% every year.
In this, I implemented a Deterministic Model to calculate the population growth over 10 years using a fixed growth rate of 10%.
